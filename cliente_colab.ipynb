# 🧪 Probar la API de Tareas desde Colab
Pon en `URL` la dirección que te hayan pasado (la de ngrok) y ejecuta las celdas.

También puedes abrir `URL/docs` en el navegador para probarla sin código.

```
import requests

URL = "https://xxxx.ngrok-free.app"   # 👈 cambia esto por la URL que te pasen
H = {"ngrok-skip-browser-warning": "true"}  # evita la página de aviso de ngrok

print(requests.get(f"{URL}/estado", headers=H).json())
```

```
# Saludar
print(requests.get(f"{URL}/saludo/Compañero", headers=H).json())
```

```
# Crear una tarea
r = requests.post(f"{URL}/tareas", json={"titulo": "Probar la API", "autor": "Mi nombre"}, headers=H)
print(r.status_code, r.json())
tid = r.json()["id"]
```

```
# Marcarla como completada
r = requests.put(f"{URL}/tareas/{tid}", json={"titulo": "Probar la API", "autor": "Mi nombre", "completada": True}, headers=H)
print(r.json())
```

```
# Ver todas las tareas (o solo las completadas: params={"completada": True})
import pandas as pd
pd.DataFrame(requests.get(f"{URL}/tareas", headers=H).json())
```

```
# Borrar la tarea
print(requests.delete(f"{URL}/tareas/{tid}", headers=H).json())
```
